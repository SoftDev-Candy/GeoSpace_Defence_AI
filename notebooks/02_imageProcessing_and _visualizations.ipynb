```python
def process_image(img_path):
    opened_image = Image.open(img_path)
    resize_img = opened_image.resize((128 , 128))
    img_array = np.array(resize_img)
    img_array_normalised = img_array/255
    return img_array_normalised

```

```python
def build_dataset(metadata , split):
    list_X = []
    list_Y = []

    split_data = metadata[metadata["split"] == split]
    for index , row in split_data.iterrows():
        image_path = (dataset_path/row["split"].lower()
              /row["class_labels"]
              /row["filename"])
        process_img = process_image(image_path)
        list_X.append(process_img)
        list_Y.append(row["class_labels"])

    list_X = np.array(list_X)
    list_Y = np.array(list_Y)

    return list_X , list_Y


```

```python
build_dataset(metadata , "Train")
```
