```python
from pathlib import Path
```

```python
import pandas as pd
from PIL import Image
```

```python
#locate the dataset
dataset_path = Path("../data/MSTAR-10-Classes")
dataset_path.exists()
```

```python
#create 2 paths and use '/' to join them next
train_path = dataset_path / "train"
train_path.exists()
```

```python
test_path  = dataset_path / "test"
#verify paths to ensure they exist to avoid further complications
test_path.exists()
```

```python
train_classes = [
    folder.name
    for folder in train_path.iterdir()
    if folder.is_dir()
]
```

```python
train_classes
```

```python
#Examine image formats to understand the data
bmp2_folder = train_path / "bmp2"
bmp2_folder.exists()

img_files = [file_path
             for file_path in bmp2_folder . iterdir()
if file_path.is_file()]

print("Number of BMP2 files : ",  len(img_files))
```

```python
for file_path in img_files[:5]:
    print(file_path.name , file_path.suffix)
```

```python
#open one image to test what it looks like and what it contains
sample_file = img_files[0]

with Image.open(sample_file) as image:
    print("File:", sample_file.name)
    print("Format : " , image.format)
    print("Size:", image.size)
    print("Mode:", image.mode)
    #L is greyscale which most SAR images are and
    #RGB is well colour
```

```python
#count files per classes each class would have a different numbers of files need to check and see the count
#Create a list of dictionaries:
class_counts = []

splits = [("Train" , train_path) , ("Test" , test_path)]

for split_name , split_path in splits:

    for class_folder in split_path.iterdir():

        if class_folder.is_dir():

            image_files = [
                file_path
                for file_path in class_folder.iterdir()
                if file_path.is_file()
               ]
            class_counts.append ({
    "split": split_name,
    "class_labels": class_folder.name,
    "img_counts": len(image_files)
})
```

```python
print(class_counts[:3])
print("Number of rows :" , len(class_counts))
```

```python
df_counts = pd.DataFrame(class_counts)
df_counts
```

```python
df_counts = df_counts.sort_values(by = ["split", "img_counts"], ascending = [True,False])
```
